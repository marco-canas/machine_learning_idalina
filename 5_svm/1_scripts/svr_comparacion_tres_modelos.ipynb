```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.patches import Patch

# -----------------------------------------------------------------------------
# CONFIGURACIÓN DE RUTAS
# -----------------------------------------------------------------------------
PATH_RESULTS = r"C:\Users\marco\Documentos\investigacion\machine_learning_idalina\5_svm\3_resultados"
os.makedirs(PATH_RESULTS, exist_ok=True)

# -----------------------------------------------------------------------------
# DATOS DE LOS TRES MODELOS
# -----------------------------------------------------------------------------

# 1. MODELO BASE (Con todos los predictores)
modelo_base_data = {
    'Split': ['80-20', '90-10', '95-5'],
    'Best_Params': [
        "{'C': 100, 'epsilon': 0.001, 'gamma': 0.0001, 'kernel': 'rbf'}",
        "{'C': 500, 'epsilon': 0.01, 'gamma': 0.0001, 'kernel': 'rbf'}",
        "{'C': 1000, 'epsilon': 0.001, 'gamma': 0.0001, 'kernel': 'rbf'}"
    ],
    'MAE_Train': [5.594, 5.559, 5.699],
    'MAE_Test': [8.494, 8.396, 7.666],
    'MAE_Promedio': [7.044, 6.977, 6.683],
    'RMSE_Train': [8.720, 8.414, 8.536],
    'RMSE_Test': [10.681, 10.745, 10.409],
    'R2_Train': [0.863, 0.868, 0.859],
    'R2_Test': [0.487, 0.283, -0.468],
    'MAE_Picos_Train': [12.160, 11.490, 11.532],
    'MAE_Picos_Test': [10.927, 8.689, 8.129],
    'MAE_Picos_Promedio': [11.543, 10.089, 9.831]
}

# 2. MODELO METEOROLÓGICO (Solo variables meteorológicas)
modelo_meteo_data = {
    'Split': ['95-5'],
    'Best_Params': ['Optimizado para MAE < 4'],
    'MAE_Train': [6.351],
    'MAE_Test': [7.217],
    'MAE_Promedio': [6.784],
    'RMSE_Train': [None],
    'RMSE_Test': [None],
    'R2_Train': [None],
    'R2_Test': [None],
    'MAE_Picos_Train': [12.583],
    'MAE_Picos_Test': [7.671],
    'MAE_Picos_Promedio': [10.127]
}

# 3. MODELO AUTOREGRESIVO (Solo rezagos de casos_dengue)
modelo_autoreg_data = {
    'Split': ['95-5'],
    'Best_Params': ['Optimizado para MAE < 4'],
    'MAE_Train': [5.610],
    'MAE_Test': [7.053],
    'MAE_Promedio': [6.331],
    'RMSE_Train': [None],
    'RMSE_Test': [None],
    'R2_Train': [None],
    'R2_Test': [None],
    'MAE_Picos_Train': [11.385],
    'MAE_Picos_Test': [7.319],
    'MAE_Picos_Promedio': [9.352]
}

# Crear DataFrames
df_base = pd.DataFrame(modelo_base_data)
df_meteo = pd.DataFrame(modelo_meteo_data)
df_autoreg = pd.DataFrame(modelo_autoreg_data)

# Seleccionar el mejor split para cada modelo
df_base_best = df_base.loc[df_base['MAE_Promedio'].idxmin()].to_frame().T
df_meteo_best = df_meteo
df_autoreg_best = df_autoreg

# DataFrame consolidado para comparación
df_comparison = pd.DataFrame({
    'Métrica': ['MAE Train', 'MAE Test', 'MAE Promedio', 
                'RMSE Train', 'RMSE Test', 
                'R2 Train', 'R2 Test',
                'MAE Picos Train', 'MAE Picos Test', 'MAE Picos Promedio'],
    'Modelo Base (Todos los predictores)': [
        df_base_best['MAE_Train'].values[0],
        df_base_best['MAE_Test'].values[0],
        df_base_best['MAE_Promedio'].values[0],
        df_base_best['RMSE_Train'].values[0],
        df_base_best['RMSE_Test'].values[0],
        df_base_best['R2_Train'].values[0],
        df_base_best['R2_Test'].values[0],
        df_base_best['MAE_Picos_Train'].values[0],
        df_base_best['MAE_Picos_Test'].values[0],
        df_base_best['MAE_Picos_Promedio'].values[0]
    ],
    'Modelo Meteorológico': [
        df_meteo_best['MAE_Train'].values[0],
        df_meteo_best['MAE_Test'].values[0],
        df_meteo_best['MAE_Promedio'].values[0],
        df_meteo_best['RMSE_Train'].values[0] if pd.notna(df_meteo_best['RMSE_Train'].values[0]) else 0,
        df_meteo_best['RMSE_Test'].values[0] if pd.notna(df_meteo_best['RMSE_Test'].values[0]) else 0,
        df_meteo_best['R2_Train'].values[0] if pd.notna(df_meteo_best['R2_Train'].values[0]) else 0,
        df_meteo_best['R2_Test'].values[0] if pd.notna(df_meteo_best['R2_Test'].values[0]) else 0,
        df_meteo_best['MAE_Picos_Train'].values[0],
        df_meteo_best['MAE_Picos_Test'].values[0],
        df_meteo_best['MAE_Picos_Promedio'].values[0]
    ],
    'Modelo Autoregresivo': [
        df_autoreg_best['MAE_Train'].values[0],
        df_autoreg_best['MAE_Test'].values[0],
        df_autoreg_best['MAE_Promedio'].values[0],
        df_autoreg_best['RMSE_Train'].values[0] if pd.notna(df_autoreg_best['RMSE_Train'].values[0]) else 0,
        df_autoreg_best['RMSE_Test'].values[0] if pd.notna(df_autoreg_best['RMSE_Test'].values[0]) else 0,
        df_autoreg_best['R2_Train'].values[0] if pd.notna(df_autoreg_best['R2_Train'].values[0]) else 0,
        df_autoreg_best['R2_Test'].values[0] if pd.notna(df_autoreg_best['R2_Test'].values[0]) else 0,
        df_autoreg_best['MAE_Picos_Train'].values[0],
        df_autoreg_best['MAE_Picos_Test'].values[0],
        df_autoreg_best['MAE_Picos_Promedio'].values[0]
    ]
})

# Guardar comparación en Excel
df_comparison.to_excel(os.path.join(PATH_RESULTS, "comparacion_tres_modelos.xlsx"), index=False)

# -----------------------------------------------------------------------------
# GRÁFICOS DE COMPARACIÓN
# -----------------------------------------------------------------------------

# Datos para gráficos
metricas_mae = ['MAE Train', 'MAE Test', 'MAE Promedio', 'MAE Picos Train', 'MAE Picos Test', 'MAE Picos Promedio']
metricas_rmse = ['RMSE Train', 'RMSE Test']
metricas_r2 = ['R2 Train', 'R2 Test']

# Extraer valores para MAE
valores_base_mae = [df_comparison[df_comparison['Métrica'] == m]['Modelo Base (Todos los predictores)'].values[0] for m in metricas_mae]
valores_meteo_mae = [df_comparison[df_comparison['Métrica'] == m]['Modelo Meteorológico'].values[0] for m in metricas_mae]
valores_autoreg_mae = [df_comparison[df_comparison['Métrica'] == m]['Modelo Autoregresivo'].values[0] for m in metricas_mae]

# Extraer valores para RMSE
valores_base_rmse = [df_comparison[df_comparison['Métrica'] == m]['Modelo Base (Todos los predictores)'].values[0] for m in metricas_rmse]
valores_meteo_rmse = [df_comparison[df_comparison['Métrica'] == m]['Modelo Meteorológico'].values[0] for m in metricas_rmse]
valores_autoreg_rmse = [df_comparison[df_comparison['Métrica'] == m]['Modelo Autoregresivo'].values[0] for m in metricas_rmse]

# Extraer valores para R2
valores_base_r2 = [df_comparison[df_comparison['Métrica'] == m]['Modelo Base (Todos los predictores)'].values[0] for m in metricas_r2]
valores_meteo_r2 = [df_comparison[df_comparison['Métrica'] == m]['Modelo Meteorológico'].values[0] for m in metricas_r2]
valores_autoreg_r2 = [df_comparison[df_comparison['Métrica'] == m]['Modelo Autoregresivo'].values[0] for m in metricas_r2]

# Configuración de colores
colors = {
    'Base': '#3498db',
    'Meteorológico': '#2ecc71',
    'Autoregresivo': '#e74c3c'
}

# -----------------------------------------------------------------------------
# GRÁFICO 1: MAE COMPARATIVO
# -----------------------------------------------------------------------------
fig1, ax1 = plt.subplots(figsize=(12, 7))

x = np.arange(len(metricas_mae))
width = 0.25

bars1 = ax1.bar(x - width, valores_base_mae, width, label='Modelo Base', color=colors['Base'], edgecolor='black', linewidth=1)
bars2 = ax1.bar(x, valores_meteo_mae, width, label='Modelo Meteorológico', color=colors['Meteorológico'], edgecolor='black', linewidth=1)
bars3 = ax1.bar(x + width, valores_autoreg_mae, width, label='Modelo Autoregresivo', color=colors['Autoregresivo'], edgecolor='black', linewidth=1)

# Línea de referencia MAE = 4
ax1.axhline(y=4, color='red', linestyle='--', linewidth=2, alpha=0.7, label='Objetivo MAE < 4')

# Etiquetas y título
ax1.set_xlabel('Métricas de Error', fontsize=12, fontweight='bold')
ax1.set_ylabel('MAE', fontsize=12, fontweight='bold')
ax1.set_title('Comparación de MAE entre Modelos', fontsize=14, fontweight='bold')
ax1.set_xticks(x)
ax1.set_xticklabels(metricas_mae, rotation=15, ha='right')
ax1.legend(loc='upper left', fontsize=10)
ax1.grid(True, alpha=0.3, linestyle=':')

# Agregar valores encima de las barras
for bars in [bars1, bars2, bars3]:
    for bar in bars:
        height = bar.get_height()
        ax1.text(bar.get_x() + bar.get_width()/2., height + 0.1,
                f'{height:.2f}', ha='center', va='bottom', fontsize=8, rotation=0)

plt.tight_layout()
plt.savefig(os.path.join(PATH_RESULTS, "comparacion_mae_modelos.png"), dpi=300, bbox_inches='tight')
plt.close()

# -----------------------------------------------------------------------------
# GRÁFICO 2: RMSE COMPARATIVO
# -----------------------------------------------------------------------------
fig2, ax2 = plt.subplots(figsize=(10, 6))

x = np.arange(len(metricas_rmse))
width = 0.25

bars1 = ax2.bar(x - width, valores_base_rmse, width, label='Modelo Base', color=colors['Base'], edgecolor='black', linewidth=1)
bars2 = ax2.bar(x, valores_meteo_rmse, width, label='Modelo Meteorológico', color=colors['Meteorológico'], edgecolor='black', linewidth=1)
bars3 = ax2.bar(x + width, valores_autoreg_rmse, width, label='Modelo Autoregresivo', color=colors['Autoregresivo'], edgecolor='black', linewidth=1)

ax2.set_xlabel('Métricas de Error', fontsize=12, fontweight='bold')
ax2.set_ylabel('RMSE', fontsize=12, fontweight='bold')
ax2.set_title('Comparación de RMSE entre Modelos', fontsize=14, fontweight='bold')
ax2.set_xticks(x)
ax2.set_xticklabels(metricas_rmse, rotation=0)
ax2.legend(loc='upper left', fontsize=10)
ax2.grid(True, alpha=0.3, linestyle=':')

for bars in [bars1, bars2, bars3]:
    for bar in bars:
        height = bar.get_height()
        ax2.text(bar.get_x() + bar.get_width()/2., height + 0.1,
                f'{height:.2f}', ha='center', va='bottom', fontsize=8, rotation=0)

plt.tight_layout()
plt.savefig(os.path.join(PATH_RESULTS, "comparacion_rmse_modelos.png"), dpi=300, bbox_inches='tight')
plt.close()

# -----------------------------------------------------------------------------
# GRÁFICO 3: R2 COMPARATIVO
# -----------------------------------------------------------------------------
fig3, ax3 = plt.subplots(figsize=(10, 6))

x = np.arange(len(metricas_r2))
width = 0.25

bars1 = ax3.bar(x - width, valores_base_r2, width, label='Modelo Base', color=colors['Base'], edgecolor='black', linewidth=1)
bars2 = ax3.bar(x, valores_meteo_r2, width, label='Modelo Meteorológico', color=colors['Meteorológico'], edgecolor='black', linewidth=1)
bars3 = ax3.bar(x + width, valores_autoreg_r2, width, label='Modelo Autoregresivo', color=colors['Autoregresivo'], edgecolor='black', linewidth=1)

# Línea de referencia R2 = 0
ax3.axhline(y=0, color='gray', linestyle='-', linewidth=1, alpha=0.5)

ax3.set_xlabel('Métricas de Bondad de Ajuste', fontsize=12, fontweight='bold')
ax3.set_ylabel('R² Score', fontsize=12, fontweight='bold')
ax3.set_title('Comparación de R² entre Modelos', fontsize=14, fontweight='bold')
ax3.set_xticks(x)
ax3.set_xticklabels(metricas_r2, rotation=0)
ax3.legend(loc='upper left', fontsize=10)
ax3.grid(True, alpha=0.3, linestyle=':')

for bars in [bars1, bars2, bars3]:
    for bar in bars:
        height = bar.get_height()
        ax3.text(bar.get_x() + bar.get_width()/2., height + (0.1 if height >= 0 else -0.1),
                f'{height:.3f}', ha='center', va='bottom' if height >= 0 else 'top', fontsize=8, rotation=0)

plt.tight_layout()
plt.savefig(os.path.join(PATH_RESULTS, "comparacion_r2_modelos.png"), dpi=300, bbox_inches='tight')
plt.close()

# -----------------------------------------------------------------------------
# GRÁFICO 4: RESÚMEN COMPACTO DE TODAS LAS MÉTRICAS (VERSIÓN CORREGIDA)
# -----------------------------------------------------------------------------
fig4, ax4 = plt.subplots(figsize=(14, 8))

# Definir todas las métricas con nombres exactos del DataFrame
all_metrics = ['MAE Train', 'MAE Test', 'MAE Promedio', 
               'MAE Picos Train', 'MAE Picos Test', 'MAE Picos Promedio',
               'RMSE Train', 'RMSE Test', 'R2 Train', 'R2 Test']

# Construir listas de valores directamente del DataFrame
all_values_base = []
all_values_meteo = []
all_values_autoreg = []

for m in all_metrics:
    # Buscar el valor en df_comparison
    val_base = df_comparison[df_comparison['Métrica'] == m]['Modelo Base (Todos los predictores)'].values[0]
    val_meteo = df_comparison[df_comparison['Métrica'] == m]['Modelo Meteorológico'].values[0]
    val_autoreg = df_comparison[df_comparison['Métrica'] == m]['Modelo Autoregresivo'].values[0]
    
    # Asegurar que los valores sean numéricos
    all_values_base.append(val_base if not isinstance(val_base, str) and not pd.isna(val_base) else 0)
    all_values_meteo.append(val_meteo if not isinstance(val_meteo, str) and not pd.isna(val_meteo) else 0)
    all_values_autoreg.append(val_autoreg if not isinstance(val_autoreg, str) and not pd.isna(val_autoreg) else 0)

# Verificar que todas las listas tengan la misma longitud
print(f"Longitud de all_metrics: {len(all_metrics)}")
print(f"Longitud de all_values_base: {len(all_values_base)}")
print(f"Longitud de all_values_meteo: {len(all_values_meteo)}")
print(f"Longitud de all_values_autoreg: {len(all_values_autoreg)}")

# Escalar R2 para visualización (multiplicar por 10 para que sea visible en la misma escala)
r2_indices = [8, 9]  # Índices de R2 Train y R2 Test
for idx in r2_indices:
    if idx < len(all_values_base):
        all_values_base[idx] = all_values_base[idx] * 10
        all_values_meteo[idx] = all_values_meteo[idx] * 10
        all_values_autoreg[idx] = all_values_autoreg[idx] * 10

# Crear nombres abreviados para el eje X
short_labels = ['MAE Train', 'MAE Test', 'MAE Prom', 
                'MAE Picos T', 'MAE Picos Test', 'MAE Picos Prom',
                'RMSE Train', 'RMSE Test', 'R2 Train', 'R2 Test']

x = np.arange(len(all_metrics))
width = 0.25

bars1 = ax4.bar(x - width, all_values_base, width, label='Modelo Base', color=colors['Base'], edgecolor='black', linewidth=1)
bars2 = ax4.bar(x, all_values_meteo, width, label='Modelo Meteorológico', color=colors['Meteorológico'], edgecolor='black', linewidth=1)
bars3 = ax4.bar(x + width, all_values_autoreg, width, label='Modelo Autoregresivo', color=colors['Autoregresivo'], edgecolor='black', linewidth=1)

ax4.set_xlabel('Métricas', fontsize=12, fontweight='bold')
ax4.set_ylabel('Valor (R² escalado x10)', fontsize=12, fontweight='bold')
ax4.set_title('Comparación Integral de Modelos - Todas las Métricas', fontsize=14, fontweight='bold')
ax4.set_xticks(x)
ax4.set_xticklabels(short_labels, rotation=45, ha='right')
ax4.legend(loc='upper left', fontsize=10)
ax4.grid(True, alpha=0.3, linestyle=':')

# Agregar valores encima de las barras
for bars in [bars1, bars2, bars3]:
    for bar in bars:
        height = bar.get_height()
        ax4.text(bar.get_x() + bar.get_width()/2., height + 0.2,
                f'{height:.1f}', ha='center', va='bottom', fontsize=7, rotation=0)

# Agregar nota sobre R² escalado
ax4.text(0.98, 0.02, 'Nota: R² escalado x10 para visualización', 
         transform=ax4.transAxes, fontsize=8, verticalalignment='bottom',
         horizontalalignment='right', bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5))

plt.tight_layout()
plt.savefig(os.path.join(PATH_RESULTS, "comparacion_integral_modelos.png"), dpi=300, bbox_inches='tight')
plt.close()

# -----------------------------------------------------------------------------
# GENERAR REPORTE DE RESULTADOS
# -----------------------------------------------------------------------------
print("=" * 70)
print("COMPARACIÓN DE MODELOS - RESULTADOS")
print("=" * 70)
print(f"\nMejores resultados por modelo:")
print("-" * 70)
print(f"Modelo Base (Todos los predictores):")
print(f"  - Split: {df_base_best['Split'].values[0]}")
print(f"  - MAE Promedio: {df_base_best['MAE_Promedio'].values[0]:.3f}")
print(f"  - MAE Picos Promedio: {df_base_best['MAE_Picos_Promedio'].values[0]:.3f}")
print(f"  - R2 Test: {df_base_best['R2_Test'].values[0]:.3f}")

print(f"\nModelo Meteorológico:")
print(f"  - Split: {df_meteo_best['Split'].values[0]}")
print(f"  - MAE Promedio: {df_meteo_best['MAE_Promedio'].values[0]:.3f}")
print(f"  - MAE Picos Promedio: {df_meteo_best['MAE_Picos_Promedio'].values[0]:.3f}")

print(f"\nModelo Autoregresivo:")
print(f"  - Split: {df_autoreg_best['Split'].values[0]}")
print(f"  - MAE Promedio: {df_autoreg_best['MAE_Promedio'].values[0]:.3f}")
print(f"  - MAE Picos Promedio: {df_autoreg_best['MAE_Picos_Promedio'].values[0]:.3f}")

print("\n" + "=" * 70)
print("MEJOR MODELO SEGÚN MAE PROMEDIO:")
best_model = df_comparison.loc[df_comparison['Métrica'] == 'MAE Promedio']
best_value = min(best_model['Modelo Base (Todos los predictores)'].values[0],
                 best_model['Modelo Meteorológico'].values[0],
                 best_model['Modelo Autoregresivo'].values[0])

if best_value == best_model['Modelo Base (Todos los predictores)'].values[0]:
    print("✅ Modelo Base (Todos los predictores)")
elif best_value == best_model['Modelo Meteorológico'].values[0]:
    print("✅ Modelo Meteorológico")
else:
    print("✅ Modelo Autoregresivo")
print(f"   MAE Promedio: {best_value:.3f}")

print("\n" + "=" * 70)
print("ARCHIVOS GENERADOS:")
print(f"  - {os.path.join(PATH_RESULTS, 'comparacion_tres_modelos.xlsx')}")
print(f"  - {os.path.join(PATH_RESULTS, 'comparacion_mae_modelos.png')}")
print(f"  - {os.path.join(PATH_RESULTS, 'comparacion_rmse_modelos.png')}")
print(f"  - {os.path.join(PATH_RESULTS, 'comparacion_r2_modelos.png')}")
print(f"  - {os.path.join(PATH_RESULTS, 'comparacion_integral_modelos.png')}")
print("=" * 70)
```
