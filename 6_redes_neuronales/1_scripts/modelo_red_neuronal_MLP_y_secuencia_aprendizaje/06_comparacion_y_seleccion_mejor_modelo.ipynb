## Objetivo

Construir una tabla comparativa con los resultados de todas las MLP.

El modelo ganador no debe seleccionarse únicamente por el menor MAE
de entrenamiento.

Se debe considerar especialmente su comportamiento sobre `test`.


```python

import pandas as pd
import numpy as np
```

## 1. Tabla comparativa

```python

# Ejemplo de estructura esperada

resultados = [

    {
        "metodo": "importancia",
        "n_features": 3,
        "mae_train": 0.0,
        "mae_test": 0.0,
        "rmse_train": 0.0,
        "rmse_test": 0.0,
        "r2_train": 0.0,
        "r2_test": 0.0
    }

]

df_resultados = pd.DataFrame(
    resultados
)

df_resultados
```

## 2. Selección

```python

def seleccionar_mejor_modelo(
    resultados
):

    return min(
        resultados,
        key=lambda r: (
            not r["cumple_objetivo"],
            max(
                r["mae_train"],
                r["mae_test"]
            )
        )
    )
```


### Criterio

El criterio utilizado en la consolidación prioriza:

1. cumplir el objetivo de MAE;
2. mantener bajo el peor MAE entre train y test.

Esto evita seleccionar simplemente un modelo que memorice los datos
de entrenamiento.
