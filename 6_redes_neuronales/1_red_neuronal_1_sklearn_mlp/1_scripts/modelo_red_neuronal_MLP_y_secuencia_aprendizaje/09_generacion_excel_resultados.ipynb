
# 09. Generación de archivos Excel

## Objetivo

Estudiar la etapa final de consolidación de resultados.

El proceso genera archivos Excel para documentar:

- comparación de modelos;
- predicciones;
- métricas;
- características utilizadas;
- información del mejor modelo.


```python

import pandas as pd
from datetime import datetime
```

## 1. Excel comparativo

```python

timestamp = datetime.now().strftime(
    "%Y%m%d_%H%M%S"
)

archivo_comparativa = (
    RUTA_PROCESADOS /
    f"resultados_mlp_comparativa_{timestamp}.xlsx"
)

df_comparativa.to_excel(
    archivo_comparativa,
    index=False
)

print(
    "Comparativa guardada:",
    archivo_comparativa
)
```

## 2. Predicciones del mejor modelo

```python

df_pred_train = pd.DataFrame({

    "fecha": df_train_fecha,

    "casos_dengue_real":
        y_train_real,

    "casos_dengue_predicho":
        pred_train_real,

    "conjunto":
        "train"
})


df_pred_test = pd.DataFrame({

    "fecha": df_test_fecha,

    "casos_dengue_real":
        y_test_real,

    "casos_dengue_predicho":
        pred_test_real,

    "conjunto":
        "test"
})


df_predicciones = pd.concat(
    [
        df_pred_train,
        df_pred_test
    ],
    ignore_index=True
)

archivo_predicciones = (
    RUTA_PROCESADOS /
    f"mlp_mejor_modelo_predicciones_{timestamp}.xlsx"
)

df_predicciones.to_excel(
    archivo_predicciones,
    index=False
)

print(
    "Predicciones guardadas:",
    archivo_predicciones
)
```
