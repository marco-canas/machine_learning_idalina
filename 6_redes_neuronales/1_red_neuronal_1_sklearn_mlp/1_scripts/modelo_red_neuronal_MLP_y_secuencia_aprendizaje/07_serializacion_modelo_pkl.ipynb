Estudiar cómo convertir el modelo entrenado en un archivo `.pkl`

El archivo debe contener:

- MAE de prueba.

## 2. Guardar el modelo

```python

timestamp = datetime.now().strftime(
    "%Y%m%d_%H%M%S"
)

archivo_modelo = (
    RUTA_PROCESADOS /
    f"mlp_mejor_modelo_{metodo}_{timestamp}.pkl"
)

joblib.dump(
    objeto_modelo,
    archivo_modelo
)

print(
    "Modelo guardado en:",
    archivo_modelo
)
```


### Idea fundamental

El `.pkl` no contiene únicamente la red neuronal.

Contiene el conjunto necesario para reproducir el flujo de predicción,
incluyendo los escaladores utilizados durante el entrenamiento.
