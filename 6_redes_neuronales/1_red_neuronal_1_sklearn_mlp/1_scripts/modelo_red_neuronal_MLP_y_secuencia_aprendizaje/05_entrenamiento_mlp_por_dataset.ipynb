```python

import numpy as np
import pandas as pd

from sklearn.neural_network import MLPRegressor
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)
```

## 1. Métricas

```python

def calcular_metricas(
    y_real,
    y_pred
):

    return {

        "MAE": mean_absolute_error(
            y_real,
            y_pred
        ),

        "RMSE": np.sqrt(
            mean_squared_error(
                y_real,
                y_pred
            )
        ),

        "R2": r2_score(
            y_real,
            y_pred
        )
    }
```

## 2. Entrenamiento

```python

def entrenar_mlp(
    X_train,
    y_train,
    X_test,
    y_test,
    params
):

    modelo = MLPRegressor(
        **params,
        max_iter=3000,
        early_stopping=True,
        random_state=42
    )

    modelo.fit(
        X_train,
        y_train
    )

    pred_train = modelo.predict(
        X_train
    )

    pred_test = modelo.predict(
        X_test
    )

    metricas_train = calcular_metricas(
        y_train,
        pred_train
    )

    metricas_test = calcular_metricas(
        y_test,
        pred_test
    )

    return (
        modelo,
        pred_train,
        pred_test,
        metricas_train,
        metricas_test
    )
```
