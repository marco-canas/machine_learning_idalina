
# 08. Dashboard del modelo predictor

## Objetivo

Construir un dashboard para diagnosticar visualmente el comportamiento
del modelo seleccionado.

El dashboard permite analizar:

- MAE por método;
- R² por método;
- valores reales frente a predichos;
- comportamiento temporal;
- distribución de residuales.


```python

import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

## 1. Comparación de modelos

```python

def dashboard_comparacion(df):

    fig, axes = plt.subplots(
        1,
        2,
        figsize=(14, 5)
    )

    axes[0].bar(
        df["Método"],
        df["MAE test"]
    )

    axes[0].set_title(
        "MAE de los modelos"
    )

    axes[0].set_ylabel(
        "MAE"
    )

    axes[1].bar(
        df["Método"],
        df["R2 test"]
    )

    axes[1].set_title(
        "R² de los modelos"
    )

    axes[1].set_ylabel(
        "R²"
    )

    plt.tight_layout()

    return fig
```

## 2. Real frente a predicho

```python

def grafica_real_predicho(
    fechas,
    y_real,
    y_pred
):

    fig, ax = plt.subplots(
        figsize=(14, 5)
    )

    ax.plot(
        fechas,
        y_real,
        label="Real"
    )

    ax.plot(
        fechas,
        y_pred,
        label="Predicho"
    )

    ax.set_title(
        "Casos reales frente a predichos"
    )

    ax.set_xlabel("Fecha")
    ax.set_ylabel("Casos de dengue")

    ax.legend()

    ax.grid(
        alpha=0.3
    )

    plt.tight_layout()

    return fig
```

## 3. Residuales

```python

def grafica_residuales(
    y_real,
    y_pred
):

    residuales = (
        np.asarray(y_real) -
        np.asarray(y_pred)
    )

    fig, ax = plt.subplots(
        figsize=(10, 5)
    )

    ax.hist(
        residuales,
        bins=30
    )

    ax.axvline(
        0,
        linestyle="--"
    )

    ax.set_title(
        "Distribución de residuales"
    )

    ax.set_xlabel(
        "Residual"
    )

    ax.set_ylabel(
        "Frecuencia"
    )

    plt.tight_layout()

    return fig
```
