
# 04. Búsqueda de hiperparámetros

## Objetivo

Estudiar cómo se seleccionan los hiperparámetros de `MLPRegressor`.

La estrategia de consolidación utiliza:

- `RandomizedSearchCV`
- `TimeSeriesSplit`
- diferentes arquitecturas;
- funciones de activación;
- `alpha`;
- `learning_rate_init`;
- `batch_size`.


```python

import numpy as np

from scipy.stats import loguniform

from sklearn.neural_network import MLPRegressor
from sklearn.model_selection import (
    RandomizedSearchCV,
    TimeSeriesSplit
)
```

## 1. Modelo base

```python

modelo = MLPRegressor(
    max_iter=3000,
    early_stopping=True,
    random_state=42
)
```

## 2. Espacio de búsqueda

```python

param_distributions = {

    "hidden_layer_sizes": [
        (64, 32),
        (64, 32, 16),
        (128, 64, 32),
        (128, 64, 32, 16),
        (256, 128, 64, 32)
    ],

    "activation": [
        "relu",
        "tanh"
    ],

    "alpha": loguniform(
        1e-5,
        1e-1
    ),

    "learning_rate_init": loguniform(
        1e-4,
        1e-2
    ),

    "batch_size": [
        16,
        32,
        64
    ]
}
```

## 3. Validación temporal

```python

tscv = TimeSeriesSplit(
    n_splits=3
)

print(tscv)
```


La utilización de `TimeSeriesSplit` mantiene la estructura temporal
durante la validación cruzada y evita mezclar información futura con
información pasada.
