```python
from sklearn.datasets import load_iris 
```

```python
iris = load_iris(as_frame=True)
```

```python
X = iris.data.iloc[:, 2:]  # we only take the third to fourth features.
y = iris.target
```

```python
from sklearn.tree import DecisionTreeClassifier
iris_clf = DecisionTreeClassifier(max_depth = 2, random_state = 42)
iris_clf.fit(X, y)
```

```python
import pandas as pd 
df_width = pd.concat([pd.DataFrame(X), pd.DataFrame(y)], axis=1)
```

```python
X.columns
```

```python
df_width.sort_values(by=['petal width (cm)'], ascending=True, inplace=True)
```

```python
df_length = iris.data.iloc[:, 2:3].sort_values(by=['petal length (cm)'], ascending=True, inplace=True)
```

```python
import matplotlib.pyplot as plt

plt.plot(df_width['petal width (cm)'][y == 0], df_width['petal length (cm)'][y == 0], 'o', label='Setosa')

plt.plot(df_width['petal width (cm)'][y == 1], df_width['petal length (cm)'][y == 1], 'o', label='Versicolor')
plt.grid(alpha=0.3)
plt.legend()
plt.xlabel('Petal Width (cm)')
plt.ylabel('Petal Length (cm)')
plt.hlines(y=2.45, xmin=0, xmax=2.5, color='black', linestyles='--')
plt.vlines(x=0.84, ymin=0, ymax=7, color='black', linestyles='--')

plt.savefig(r"C:\Users\marco\Downloads\decision_tree_2.png", dpi=300, bbox_inches='tight')
```
